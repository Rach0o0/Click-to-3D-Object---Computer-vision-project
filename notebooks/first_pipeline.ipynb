 # First pipeline


How it works:

1.  count_frames and read_frame read the source; the extraction cell writes frames/0.jpg … 99.jpg.
2.  imshow + on_click write click.json.
3.  build_sam2_video_predictor + init_state load the model and the clip into state.
4.  add_click turns the click into a mask on frame 0; the cell after it shows red_overlay(load_frame(0), mask0) so you can check it.
5.  propagate tracks that mask through all frames; imshow previews 5 of them.
6.  save_red writes outputs/pedestrians/red/*.png and data/pedestrians/masks/*.npy.

### IMSHOW

```python
def imshow(
    images: list[np.ndarray] | np.ndarray,
    titles: list[str] | str | None = None,
    nrows: int = 0,
    ncols: int = 0,
    figsize: (float, float) = (10, 4),
):
    """Plot a multiple images with titles.

    Parameters
    ----------
    images : image list
    titles : title list
    ncols : number of columns of subplots wanted in the display
    nrows : number of rows of subplots wanted in the figure
    figsize : size of the figure
    """
    if not isinstance(images, list):
        images = [images]
    if titles is not None and not isinstance(titles, list):
        titles = [titles]

    if ncols == 0 and nrows == 0:
        ncols = len(images)
        nrows = 1
    if ncols == 0:
        ncols = len(images) // nrows
    if nrows == 0:
        nrows = len(images) // ncols

    fig, axeslist = plt.subplots(
        ncols=ncols, nrows=nrows, squeeze=False, figsize=figsize
    )

    for i, image in enumerate(images):
        axeslist.ravel()[i].imshow(image, cmap=plt.gray(), vmin=0, vmax=255)
        if titles is not None:
            axeslist.ravel()[i].set_title(titles[i])
        axeslist.ravel()[i].set_axis_off()
    plt.tight_layout()  # optional

    return fig, axeslist
```

## Read frames

```python
#check if it is a video :
def is_video(source):
    return source.endswith((".mp4", ".mov", ".avi"))

#open a video : cv2.VideoCapture(source)
#close a video : video.release()
#read a frame : video.read()
#go at k-th frame : video.set(cv2.CAP_PROP_POS_FRAMES, k)  

#count number of frames
def count_frames(source):
    if is_video(source):
        video = cv2.VideoCapture(source)
        n = int(video.get(cv2.CAP_PROP_FRAME_COUNT))
        video.release()
        return n
    return len(os.listdir(source)) #count files if not a video

#read the k-th frame, for a video: opens file jumps to frame k and reads it and converts to RGB, 
#for a folder: reads the k-th file in the folder
def read_frame(source, k):
    if is_video(source):
        video = cv2.VideoCapture(source)
        video.set(cv2.CAP_PROP_POS_FRAMES, k)
        _, frame = video.read()
        video.release()
        return cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    #else : take the list of frames
    names = sorted(os.listdir(source))
    return io.imread(f"{source}/{names[k]}")
```

## SAM 2

### Initialiser le modèle: A run qu'une seule fois

```python
device = "cuda" if torch.cuda.is_available() else ("mps" if torch.backends.mps.is_available() else "cpu")

#Build the SAM 2 video predictor from the config and checkpoint + sets device to use
predictor = build_sam2_video_predictor(
    "configs/sam2.1/sam2.1_hiera_s.yaml", 
    CHECKPOINT, device=device)

#loads every JPEG in the folder and returns state (memory of SAM keeps for this video)
#you run it once per clip!
state = predictor.init_state(
    video_path=str(FRAMES_dir), 
    offload_video_to_cpu=True)

print("frames chargées :", state["num_frames"])     # doit afficher 100
```

### Helper Functions

```python
# This fucntions runs SAM 2 on the click and returns the mask, the click and the frame where it was made
from sympy import fps


def add_click(state, click_file=CLICK_file, obj_id=1, positive=True, reset=True):
    """Read the click from the JSON file and give it to SAM 2.
    reset=True  -> forget previous clicks (fresh start)
    reset=False -> add a correction click on top of previous ones"""
    if reset: 
        predictor.reset_state(state)

    with open(click_file) as f:
        c = json.load(f)
    click, click_frame = (c["x"], c["y"]), c["frame"]

    # Add the click to the predictor and get the mask 
    _, obj_ids, mask_logits = predictor.add_new_points_or_box(
        inference_state=state,
        frame_idx=click_frame,
        obj_id=obj_id,
        points=np.array([click], dtype=np.float32),
        labels=np.array([1 if positive else 0], dtype=np.int32),
    )
    #The logits>0 turns the logits into a binary mask (True for pixels belonging to the object, False otherwise)
    mask = (mask_logits[0, 0] > 0).cpu().numpy() 

    print(f"frame {click_frame}, click {click}: {mask.sum()} pixels in mask")
    return mask, click, click_frame

# Takes RGB frame and a binary mask, returns a copy where the masked pixels are blended 50/50 with red. 
def red_overlay(frame_rgb, mask, alpha=0.5):
    """Blend the mask pixels with red (frame in RGB)."""
    out = frame_rgb.copy()
    out[mask] = ((1 - alpha) * out[mask] + alpha * np.array([255, 0, 0])).astype(np.uint8)
    return out

# This function loads a frame from the FRAMES_dir in RGB format.
def load_frame(idx):
    """Read frame idx from FRAMES_dir, in RGB."""
    return cv2.cvtColor(cv2.imread(str(FRAMES_dir / f"{idx}.jpg")), cv2.COLOR_BGR2RGB)

# This function tracks the clicked object through all frames and returns a dictionary mapping frame indices to masks.
def propagate(state):
    """Track the clicked object through all frames. Returns {frame_idx: mask}."""
    video_masks = {}

    #The propagate function walks through each frame and uses memory of earlier frames to fund same object
    for frame_idx, obj_ids, mask_logits in predictor.propagate_in_video(state):
        video_masks[frame_idx] = (mask_logits[0, 0] > 0).cpu().numpy()
    return video_masks

# This function saves the red-highlighted frames as PNG images and the raw masks as NPY files.
def save_red(video_masks):
    """Save red-highlighted frames (PNG) + raw masks (NPY)."""
    os.makedirs(OUTPUT_dir, exist_ok=True)
    os.makedirs(MASKS_dir, exist_ok=True)
    for idx, mask in video_masks.items():
        out = red_overlay(load_frame(idx), mask)
        cv2.imwrite(f"{OUTPUT_dir}/{idx}.png", cv2.cvtColor(out, cv2.COLOR_RGB2BGR))  # cv2 saves BGR
        np.save(f"{MASKS_dir}/{idx}.npy", mask)
    print(len(video_masks), "frames saved in", OUTPUT_dir)

#Save as a video mp4
def save_video(video_masks, output_file="output.mp4", fps=30):
    """Save the red-highlighted frames as a video."""
    h, w = load_frame(0).shape[:2]
    vw = cv2.VideoWriter(f"{OUTPUT_dir}/{output_file}", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
    for i in sorted(video_masks):
        vw.write(cv2.imread(f"{OUTPUT_dir}/{i}.png"))
    vw.release()
```

## Delete object

```python
def background_plate(video_masks):
    """Median of each pixel over time, ignoring masked pixels."""
    idx = sorted(video_masks)

    #Arrays of frames and of the masks for the pedestrian
    frames = np.stack([load_frame(i) for i in idx]).astype(np.float32)
    pedestrian_masks = np.stack([video_masks[i] for i in idx])

    #for the pedestrian masked pixels, we set the value to NaN in the frames array so that they are ignored in the median computation
    frames[pedestrian_masks] = np.nan

    median_frames = np.nanmedian(frames, axis=0, overwrite_input=True)

    #hole mask is True for pixels that are NaN in the median frame (i.e. pixels that were always masked by the pedestrian)
    hole_mask = np.isnan(median_frames).any(axis=2).astype(np.uint8) * 255

    # Convert the median frame to uint8, replacing NaN with 0
    uint8_frame = np.nan_to_num(median_frames).astype(np.uint8)
    
    # Inpaint the holes in the median frame using the hole mask
    filled = cv2.inpaint(uint8_frame, hole_mask, 3, cv2.INPAINT_TELEA)
    return filled


def remove_object(frame, mask, bg, dilate=7):
    """Replace the (slightly enlarged) mask with the background plate."""
    # Dilate makes the mask slightly larger to ensure complete removal of the object. 
    # The dilated mask is then used to replace the corresponding pixels in the frame with those from the background plate.
    # The >0 ensures that the dilated mask is a boolean array, where True indicates the pixels to be replaced.
    m = cv2.dilate(mask.astype(np.uint8), np.ones((dilate, dilate), np.uint8)) > 0

    out = frame.copy()
    out[m] = bg[m]
    return out
```

```python
def remove_object_from_video(video_masks, bg, dilate=7):
    """Remove the object from all frames in video_masks using the background plate."""
    bg = background_plate(video_masks)
    removed_frames = {}
    for idx, mask in video_masks.items():
        frame = load_frame(idx)
        removed_frames[idx] = remove_object(frame, mask, bg, dilate)
    return removed_frames

removed_frames = remove_object_from_video(video_masks, bg=None, dilate=7)
save_video(removed_frames, output_file="removed_1.mp4", fps=30)
```
